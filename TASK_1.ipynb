```python
import pandas as pd
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
import os

```

```python
def extract_data(file_path):
    print("📥 Extracting data...")
    return pd.read_csv(file_path)
```

```python
def transform_data(df):
    print("🔧 Transforming data...")
```

```python
df = extract_data("/content/sample_data/dataset.csv")
```

```python
numerical_cols = df.select_dtypes(include=['int64', 'float64']).columns.tolist()
categorical_cols = df.select_dtypes(include=['object']).columns.tolist()
```

```python
numeric_pipeline = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='mean')),
    ('scaler', StandardScaler())
])

categorical_pipeline = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore'))
])
```

```python
preprocessor = ColumnTransformer(transformers=[
    ('num', numeric_pipeline, numerical_cols),
    ('cat', categorical_pipeline, categorical_cols)
])
```

```python
transformed_array = preprocessor.fit_transform(df)
```

```python
transformed_df = pd.DataFrame(
    transformed_array.toarray() if hasattr(transformed_array, "toarray") else transformed_array
)
```

```python
transformed_df.to_csv("/content/sample_data/transformed_data.csv", index=False)
print("✅ Transformed data saved!")
```
